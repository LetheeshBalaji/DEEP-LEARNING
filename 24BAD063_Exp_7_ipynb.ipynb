Part A: Dataset preparation and text preprocessing
Roll Number : 24BAD063


```python

# Experiment No. 7
# Text Preprocessing and Embedding Generation Using NLP Libraries

# Install required libraries
!pip install -q nltk gensim

import nltk
import string
import numpy as np

from nltk.corpus import brown, stopwords, wordnet
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download required datasets
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

# Enter your roll number
ROLL_NO = "YOUR_ROLL_NUMBER"
print("Roll Number:", ROLL_NO)

# Load Brown Corpus
sentences = brown.sents(categories='news')

# Display sample sentences
print("\nOriginal sample sentences:")
for sentence in sentences[:3]:
    print(" ".join(sentence))

# Initialize tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Convert POS tags to WordNet format
def get_wordnet_pos(word):
    from nltk import pos_tag
    tag = pos_tag([word])[0][1][0].upper()
    return {
        'J': wordnet.ADJ,
        'N': wordnet.NOUN,
        'V': wordnet.VERB,
        'R': wordnet.ADV
    }.get(tag, wordnet.NOUN)

# Preprocess sentences
processed_sentences = []

for sentence in sentences:
    text = " ".join(sentence).lower()
    tokens = word_tokenize(text)

    tokens = [
        word for word in tokens
        if word not in string.punctuation
        and word not in stop_words
        and any(char.isalpha() for char in word)
    ]

    tokens = [
        lemmatizer.lemmatize(word, get_wordnet_pos(word))
        for word in tokens
    ]

    if tokens:
        processed_sentences.append(tokens)

print("\nPreprocessed sample sentences:")
for sentence in processed_sentences[:3]:
    print(sentence)
```

Part B: Vocabulary generation and word frequency analysis
Roll Number : 24BAD063


```python

from nltk import FreqDist
from collections import Counter

# Flatten all preprocessed sentences into one token list
all_words = [
    word
    for sentence in processed_sentences
    for word in sentence
]

# Create vocabulary and frequency distribution
vocabulary = set(all_words)
freq_dist = FreqDist(all_words)

print("Vocabulary size:", len(vocabulary))
print("Total word occurrences:", len(all_words))

print("\nTop 10 most frequent words:")
for word, frequency in freq_dist.most_common(10):
    print(word, ":", frequency)

print("\nVocabulary sample:")
print(list(sorted(vocabulary))[:20])

# Prepare sentences for Word2Vec
print("\nNumber of sentences for Word2Vec:",
      len(processed_sentences))
```

Part C: Word embedding generation using Word2Vec
The lab manual specifies Skip-gram, embedding dimension 100, and context window 5.
Roll Number : 24BAD063

```python

from gensim.models import Word2Vec

# Train Word2Vec using Skip-gram
w2v_model = Word2Vec(
    sentences=processed_sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=1,
    workers=4,
    epochs=10,
    seed=42
)

print("Word2Vec model trained successfully!")

# Select a word available in the trained vocabulary
selected_word = (
    "government"
    if "government" in w2v_model.wv
    else w2v_model.wv.index_to_key[0]
)

# Display embedding vector
vector = w2v_model.wv[selected_word]

print("\nSelected word:", selected_word)
print("Embedding vector:")
print(vector)
print("Embedding dimensions:", vector.shape)

# Find five most similar words
print("\nFive most similar words:")
for word, similarity in w2v_model.wv.most_similar(
    selected_word, topn=5
):
    print(word, ":", round(similarity, 4))
```

Part D: Word similarity analysis
24BAD063

```python

# Select three meaningful word pairs
candidate_pairs = [
    ("government", "state"),
    ("city", "town"),
    ("president", "government")
]

print("Cosine similarity analysis:")

valid_scores = []

for word1, word2 in candidate_pairs:
    if word1 in w2v_model.wv and word2 in w2v_model.wv:
        score = w2v_model.wv.similarity(word1, word2)
        valid_scores.append((word1, word2, score))
        print(
            f"{word1} - {word2}: {score:.4f}"
        )
    else:
        print(
            f"{word1} - {word2}: "
            "One or both words are not in the vocabulary."
        )

if valid_scores:
    most_similar = max(valid_scores, key=lambda item: item[2])
    print("\nMost similar pair:",
          most_similar[0], "-", most_similar[1])
    print("Similarity score:", round(most_similar[2], 4))

# Save trained model
w2v_model.save("word2vec_model.model")
print("\nModel saved successfully.")
```
